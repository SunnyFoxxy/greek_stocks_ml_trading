We can easily drop all unmatching lines in both tables, in the first case its low trading company an the second - not trading at all

Merge 2 DF in one

```python
df_tnk = df_tnk.reset_index(drop=True)
df_ts = df_ts.reset_index(drop=True)

df = pd.merge(df_tnk, df_ts, left_on='symbol', right_on='symbol_en', how='inner').drop(columns=['symbol_en'])
df.head()
```

Make a new group

```python
df['liquidity_group'] = df['tick_size'].str.extract(r'(\d)').astype(int)
df.head()
```

Remove all low trading companies

```python
df = df[(~df['tick_size'].str.contains('ETF')) & (df['liquidity_group'] >= 3)].copy()
len(df)
```

Save to csv

Convert to YFinance Ticker

```python
df['yf_ticker'] = df['symbol'] + '.AT'
df.head()
```

```python
df.to_csv('candidates.csv', index=False)
```

```python
df_tickers = pd.read_csv('candidates.csv')
stocks = []
stocks_not_downloaded = []
for t in df_tickers['yf_ticker']:
    try:
        df = yf.download(t, start='2016-01-01', end='2026-11-16', progress=False, auto_adjust=False)
    except Exception as e:
        stocks_not_downloaded.append((t, e))
        continue
    if df.empty:
        stocks_not_downloaded.append((t, "no info"))
        continue
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = df.columns.get_level_values(0)


    df['Ticker'] = t
    stocks.append(df)
```

```python
df_total = pd.concat(stocks)
df_total.to_parquet('prices_raw.parquet')
```

```python
df_total = df_total.reset_index()
df_total['turnover'] = df_total['Close'] * df_total['Volume']
df_total.head()
```

```python
df_stat = df_total.groupby('Ticker', as_index=False).agg(
    first_date=('Date', 'min'),
    last_date=('Date', 'max'),
    count_dates = ('Date', 'count'),
    nill_dates = ('Volume', lambda x: (x == 0).sum()),
    median_turnover = ('turnover', 'median')
)
df_stat.sort_values('median_turnover')
```

## Stock selection criteria

Starting list: 63 stocks from Euronext Athens (company list as of 16.11.2026) with liquidity groups 3–6 (80+ trades per day, MiFID II tick size file), ETFs excluded.

Final selection from the yfinance price data (2016-01-01 to 2026-11-13, the last trading day before the download end date 2026-11-16):

| Criterion | Threshold | Why |
| --- | --- | --- |
| Price history | at least 1,800 trading days (listed in 2019 or earlier) | enough data to train and test the models |
| Median daily turnover (Close × Volume) | at least 50,000 EUR | the stock can be bought and sold without moving the price |
| Days with zero volume | not more than 3.5% of all days | the stock is traded regularly, not only from time to time |

Result: **33 stocks**.

Notes:
- ALPHA and MTLN are excluded for now: their history in yfinance starts only in 2025 because of ticker changes after company restructuring.
- The list contains only stocks that are traded today, so the results can be too optimistic (survivorship bias).

```python
df_stat = df_stat[(df_stat['median_turnover'] > 50000) & ((df_stat['count_dates'] - df_stat['nill_dates']) > 1800) & (df_stat['nill_dates']/2702 < 0.035)]
len(df_stat)
```

```python
sector_list = []
for str_stat in df_stat['Ticker']:
   sector = yf.Ticker(str_stat).info.get('sector')
   sector_list.append(sector)
df_stat['Sector'] = sector_list
df_stat.head()
```
